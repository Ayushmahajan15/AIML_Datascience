```python
import pandas as pd 
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})
```

```python
employees 
```

```python
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
departments
```

```python
pd.merge(employees,departments,on="DeptID")#it give the common deptID and show all the column in both table
```

```python
pd.merge(employees,departments,on="DeptID", how="left")#Keeps all employees, fills NaN where no match.
```

```python
pd.merge(employees,departments,on="DeptID", how="right")#right join
```

```python
pd.merge(employees,departments,on="DeptID", how="outer")#Includes all data, fills missing with NaN
```
