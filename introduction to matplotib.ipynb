```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
```

```python
plt.plot(years,runs)
```

```python
plt.plot(years,runs)
plt.xlabel("year")
plt.ylabel("Runs Scored")
plt.title("sachin Yearly Runs")
```

```python
kohli = [1,300,400,500,550,600,700,800,900,1100]
sehwag = [0,300,800,1200,1300,1500,1700,1800,1400,2000]
```

```python
plt.plot(years,kohli,label="kohli")
plt.plot(years,sehwag,label="sehwag")  
plt.xlabel("year")
plt.ylabel("Runs scored")
plt.legend()
```

```python
plt.plot(years,kohli,'r>-',label="kohli")#red triangle with solid line
plt.plot(years,sehwag,'y^:',label="sehwag")
plt.ledend()
```

```python
plt.plot(years, kohli, color='orange', linestyle='--', label="Kohli")
plt.plot(years, sehwag, color='green', linestyle='-.', label="Sehwag")
plt.plot(years, runs, color='blue', label="Tendulkar")
plt.grid(True)
plt.legend()
```
